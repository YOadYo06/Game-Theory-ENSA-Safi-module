## 1. 20 jeux aléatoires, tailles 2 à 4

Nombre d'équilibres trouvés et temps de calcul en fonction de la taille
(le nombre de couples de supports est exponentiel).

```python
df = protocole_jeux_aleatoires(tailles=(2, 3, 4), n_jeux=20, seed=0)
df.groupby('taille').agg(
    n_equilibres_moyen=('n_equilibres', 'mean'),
    temps_moyen_s=('temps_s', 'mean'),
    temps_max_s=('temps_s', 'max'),
)
```

```python
# Graphe : temps de calcul en fonction de la taille (échelle log)
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))
for taille, groupe in df.groupby('taille'):
    ax.scatter([taille] * len(groupe), groupe['temps_s'], alpha=0.6, label=f'taille {taille}')
ax.set_yscale('log')
ax.set_xlabel('taille (m = n)')
ax.set_ylabel('temps de calcul (s, log)')
ax.set_title('Énumération des supports : coût en fonction de la taille')
ax.legend()
plt.tight_layout()
plt.show()
```
